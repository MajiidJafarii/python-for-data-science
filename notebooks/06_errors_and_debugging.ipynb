<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">خطا بخشی از برنامه‌نویسی است</h2>
  <p><code>SyntaxError</code> به ساختار کد، <code>NameError</code> به نام ناشناخته، <code>TypeError</code> به نوع نامناسب و <code>ValueError</code> به مقدار نامعتبر مربوط است. آخرین خط traceback معمولاً مهم‌ترین سرنخ است.</p>
</div>

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">try و except</h2>
  <p>فقط خطایی را بگیرید که انتظارش را دارید. گرفتن همهٔ خطاها با <code>except:</code> عیب اصلی برنامه را پنهان می‌کند.</p>
</div>

```python
# Convert user-like text safely
raw_value = "12.5"

try:
    value = float(raw_value)
except ValueError:
    value = None
    print("Invalid numeric value")
else:
    print("Converted value:", value)
finally:
    print("Conversion attempt finished")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">اعتبارسنجی با raise</h2>
  <p>اگر ورودی قانون برنامه را نقض کند، با خطای مشخص اجرای نادرست را متوقف کنید.</p>
</div>

```python
# Validate a discount percentage
def apply_discount(price, percent):
    if price < 0:
        raise ValueError("price cannot be negative")
    if not 0 <= percent <= 100:
        raise ValueError("percent must be between 0 and 100")
    return price * (1 - percent / 100)

print(apply_discount(500_000, 15))

try:
    apply_discount(500_000, 130)
except ValueError as error:
    print("Validation error:", error)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">روش سادهٔ اشکال‌زدایی</h2>
  <ol><li>پیام خطا و شمارهٔ خط را کامل بخوانید.</li><li>نوع و مقدار متغیرهای همان بخش را بررسی کنید.</li><li>مسئله را با کوچک‌ترین ورودی بازتولید کنید.</li><li>یک فرض را در هر مرحله آزمایش کنید.</li></ol>
</div>

```python
# Inspect intermediate values while debugging
quantities = [2, "3", 1]
normalized = []

for item in quantities:
    print("value:", repr(item), "type:", type(item).__name__)
    normalized.append(int(item))

print("Total quantity:", sum(normalized))
```

  <h2 style="color:#173f5f;">تمرین کوتاه</h2>
  <div style="background:#fff7df; border:1px solid #f0c36d; padding:12px; border-radius:8px;">فهرست <code>['120000', 'bad', '', None, '85000.5']</code> را به عدد تبدیل کنید؛ مقدارهای نامعتبر را رد کنید و تعداد خطاها و مجموع مقدارهای معتبر را گزارش دهید.</div>
</div>

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Parse mixed price values
raw_prices = ["120000", "bad", "", None, "85000.5"]
valid_prices = []
error_count = 0

for raw_price in raw_prices:
    try:
        price = float(raw_price)
        if price < 0:
            raise ValueError("negative price")
    except (TypeError, ValueError):
        error_count += 1
        continue
    valid_prices.append(price)

print("Valid prices:", valid_prices)
print("Errors:", error_count)
print(f"Total: {sum(valid_prices):,.1f}")
```
