<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">انواع دادهٔ اصلی</h2>
  <p>چهار نوع پراستفاده عبارت‌اند از <code>int</code> برای عدد صحیح، <code>float</code> برای اعشاری، <code>str</code> برای متن و <code>bool</code> برای درست/نادرست. مقدار <code>None</code> نیز نبودن مقدار را نشان می‌دهد.</p>
</div>

```python
# Inspect common Python data types
age = 24
average = 18.75
username = "sara"
has_access = False
middle_name = None

values = [age, average, username, has_access, middle_name]
for value in values:
    print(value, "->", type(value).__name__)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">تبدیل نوع و f-string</h2>
  <p>ورودی کاربر معمولاً متن است؛ بنابراین پیش از محاسبه باید آن را تبدیل کنیم. f-string ترکیب مقدارها با متن را خواناتر می‌کند.</p>
</div>

```python
# Simulate text received from an input field
age_text = "24"
age_number = int(age_text)
next_year_age = age_number + 1

name = "سارا"
print(f"{name} سال آینده {next_year_age} ساله می‌شود.")
print(isinstance(age_number, int))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Calculate an invoice total
unit_price = 125_000
quantity = 4
tax_rate = 0.09

subtotal = unit_price * quantity
tax = subtotal * tax_rate
final_total = subtotal + tax

print(f"Subtotal: {subtotal:,.0f}")
print(f"Tax: {tax:,.0f}")
print(f"Final total: {final_total:,.0f}")
```
