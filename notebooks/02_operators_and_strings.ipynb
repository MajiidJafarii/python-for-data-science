  <h1 style="color:#173f5f; margin-bottom:4px;">نوت‌بوک 02: عملگرها و رشته‌ها</h1>
  <p><b>سطح:</b> مقدماتی &nbsp; | &nbsp; <b>روش مطالعه:</b> توضیح کوتاه ← اجرای مثال ← تغییر مثال ← حل تمرین</p>
  <h3>هدف‌های یادگیری</h3>
  <ul><li>استفاده از عملگرهای محاسباتی، مقایسه‌ای و منطقی</li>
<li>دسترسی به نویسه‌ها با اندیس و برش</li>
<li>کار با متدهای مهم رشته</li>
<li>ساخت متن استاندارد و تمیز</li></ul>
  <p style="background:#eef6fb; padding:10px; border-radius:8px;">همهٔ سلول‌ها را به‌ترتیب اجرا کنید. برای یادگیری بهتر، مقدار ورودی‌ها را تغییر دهید و نتیجه را پیش‌بینی کنید.</p>
</div>

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">عملگرهای محاسباتی</h2>
  <p>علاوه بر چهار عمل اصلی، <code>//</code> تقسیم صحیح، <code>%</code> باقی‌مانده و <code>**</code> توان را محاسبه می‌کنند.</p>
</div>

```python
# Explore arithmetic operators
total_minutes = 145
hours = total_minutes // 60
minutes = total_minutes % 60

print(hours,"," , minutes)
print(2 ** 5)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">مقایسه و منطق</h2>
  <p>نتیجهٔ مقایسه همیشه Boolean است. برای ترکیب شرط‌ها از <code>and</code>، <code>or</code> و <code>not</code> استفاده می‌شود.</p>
</div>

```python
# Combine comparison results
age = 22
has_ticket = True
is_allowed = age >= 18 and has_ticket

print(age == 22)
print(age != 18)
print(is_allowed)
print(not is_allowed)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">اندیس و برش رشته</h2>
  <p>اندیس از صفر آغاز می‌شود. اندیس منفی از انتهای رشته می‌شمارد. در برش، کران انتهایی وارد نتیجه نمی‌شود.</p>
</div>

```python
# Read parts of a string
language = "Python"

print(language[0])
print(language[-1])
print(language[0:3])
print(language[::-1])
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">متدهای پرکاربرد رشته</h2>
  <p>رشته‌ها تغییرناپذیرند؛ متدها یک رشتهٔ جدید برمی‌گردانند. برای پاک‌کردن فاصله، تغییر حروف و جایگزینی از متدهای زیر استفاده می‌کنیم.</p>
</div>

```python
# Clean and normalize text
raw_name = "  Python Data Course  "
clean_name = raw_name.strip().lower().replace(" ", "-")

print(clean_name)
print(clean_name.startswith("python"))
print(clean_name.count("a"))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">جداکردن و چسباندن</h2>
  <p><code>split</code> متن را به فهرست تبدیل می‌کند و <code>join</code> اعضای متنی یک فهرست را به هم می‌چسباند.</p>
</div>

```python
# Split a line and join selected parts
tags_text = "python, data, beginner, practice"
tags = [tag.strip() for tag in tags_text.split(",")]
normalized_tags = " | ".join(tags)

print(tags)
print(normalized_tags)
```
