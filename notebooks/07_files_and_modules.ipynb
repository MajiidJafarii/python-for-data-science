<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">مسیرها با pathlib</h2>
  <p><code>Path</code> ساخت مسیر را در سیستم‌عامل‌های مختلف خوانا و مطمئن می‌کند. در این نوت‌بوک فایل‌های نمونه در یک پوشهٔ موقت ساخته می‌شوند.</p>
</div>

```python
# Create an isolated temporary workspace
from pathlib import Path
from tempfile import TemporaryDirectory

temp_dir = TemporaryDirectory()
workspace = Path(temp_dir.name)

print("Workspace exists:", workspace.exists())
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">فایل متنی</h2>
  <p>برای متن فارسی همیشه <code>encoding='utf-8'</code> را مشخص کنید. متدهای <code>write_text</code> و <code>read_text</code> برای فایل‌های کوچک ساده‌اند.</p>
</div>

```python
# Write and read a UTF-8 text file
notes_path = workspace / "notes.txt"
notes_path.write_text("خط اول\nخط دوم\n", encoding="utf-8")

content = notes_path.read_text(encoding="utf-8")
lines = content.splitlines()

print(content)
print("Line count:", len(lines))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">JSON</h2>
  <p>JSON برای ذخیرهٔ داده‌های ساختاری مانند دیکشنری و لیست مناسب است. گزینهٔ <code>ensure_ascii=False</code> حروف فارسی را خوانا نگه می‌دارد.</p>
</div>

```python
# Save and load structured JSON data
import json

profile = {"name": "سارا", "skills": ["Python", "Excel"], "active": True}
json_path = workspace / "profile.json"

json_path.write_text(
    json.dumps(profile, ensure_ascii=False, indent=2),
    encoding="utf-8",
)
loaded_profile = json.loads(json_path.read_text(encoding="utf-8"))
print(loaded_profile)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">CSV</h2>
  <p>CSV برای داده‌های جدولی رایج است. <code>DictWriter</code> و <code>DictReader</code> هر ردیف را مانند دیکشنری مدیریت می‌کنند.</p>
</div>

```python
# Write and read a small CSV table
import csv

rows = [
    {"name": "Ali", "score": 17},
    {"name": "Sara", "score": 19},
]
csv_path = workspace / "scores.csv"

with csv_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with csv_path.open("r", newline="", encoding="utf-8") as file:
    loaded_rows = list(csv.DictReader(file))

print(loaded_rows)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ماژول‌ها</h2>
  <p>ماژول فایلی از کد قابل استفادهٔ مجدد است. کتابخانهٔ استاندارد ابزارهای آماده‌ای مانند تاریخ، آمار و مسیرها ارائه می‌کند.</p>
</div>

```python
# Use selected standard-library tools
from datetime import date, timedelta
from statistics import mean

today = date.today()
deadline = today + timedelta(days=14)
scores = [17, 19, 16.5, 18]

print("Deadline:", deadline.isoformat())
print("Average score:", mean(scores))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Persist and filter project tasks
project = {
    "name": "Website",
    "status": "active",
    "tasks": [
        {"title": "Design", "done": True},
        {"title": "Develop", "done": False},
        {"title": "Test", "done": False},
    ],
}
project_path = workspace / "project.json"
project_path.write_text(
    json.dumps(project, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

loaded_project = json.loads(project_path.read_text(encoding="utf-8"))
pending = [task["title"] for task in loaded_project["tasks"] if not task["done"]]
print("Pending tasks:", pending)

# Release the temporary workspace
temp_dir.cleanup()
```
