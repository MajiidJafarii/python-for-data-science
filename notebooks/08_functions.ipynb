  <h2 style="color:#173f5f;">چرا تابع؟</h2>
  <p>تابع یک کار مشخص را با نامی روشن انجام می‌دهد. تابع خوب ورودی و خروجی مشخص دارد و از تکرار کد جلوگیری می‌کند.</p>

```python
# Define and call a focused function
def calculate_total(price, quantity):
    return price * quantity

order_total = calculate_total(125_000, 3)
print(f"Order total: {order_total:,}")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پارامتر پیش‌فرض و آرگومان کلیدی</h2>
  <p>مقدار پیش‌فرض برای حالت رایج مناسب است. ارسال آرگومان با نام، خوانایی فراخوانی را بیشتر می‌کند.</p>
</div>

```python
# Use default and keyword arguments
def final_price(price, discount_percent=0, tax_percent=0):
    discounted = price * (1 - discount_percent / 100)
    return discounted * (1 + tax_percent / 100)

print(final_price(500_000))
print(final_price(500_000, discount_percent=10, tax_percent=9))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">چند خروجی، type hint و docstring</h2>
  <p>Type hint قرارداد ورودی و خروجی را روشن می‌کند اما نوع را در زمان اجرا اجبار نمی‌کند. در واقع چند خروجی به شکل یک تاپل بازگردانده می‌شوند.</p>
</div>

```python
# Document inputs and return multiple summary values
def summarize_scores(scores: list[float]) -> tuple[float, float, float]:
    """Return the minimum, maximum, and average score."""
    if not scores:
        raise ValueError("scores cannot be empty")
    return min(scores), max(scores), sum(scores) / len(scores)

minimum, maximum, average = summarize_scores([16, 18.5, 19, 14])
print(minimum, maximum, round(average, 2))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">*args و **kwargs</h2>
  <p>وقتی تعداد آرگومان‌ها متغیر است، <code>*args</code> آن‌ها را در تاپل و <code>**kwargs</code> آرگومان‌های نام‌دار را در دیکشنری قرار می‌دهد. فقط در صورت نیاز از آن‌ها استفاده کنید.</p>
</div>

```python
# Accept a variable number of numeric values
def average_of(*values):
    if not values:
        return 0
    return sum(values) / len(values)

# Build a readable text record from named fields
def format_record(**fields):
    return " | ".join(f"{key}={value}" for key, value in fields.items())

print(average_of(12, 15, 18))
print(format_record(name="Sara", role="Analyst", active=True))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">محدودهٔ متغیر</h2>
  <p>متغیر ساخته‌شده داخل تابع محلی است. بهتر است تابع به‌جای تغییر متغیرهای سراسری، نتیجه را برگرداند.</p>
</div>

```python
# Prefer explicit input and output over global mutation
tax_rate = 0.09

def add_tax(amount, rate):
    taxed_amount = amount * (1 + rate)
    return taxed_amount

print(add_tax(1_000_000, tax_rate))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Summarize orders above a minimum amount
def summarize_orders(orders: list[float], minimum: float = 0) -> dict:
    valid_orders = [amount for amount in orders if amount >= minimum]
    total = sum(valid_orders)
    average = total / len(valid_orders) if valid_orders else 0
    return {
        "count": len(valid_orders),
        "total": total,
        "average": average,
    }

summary = summarize_orders([250_000, 90_000, 480_000, 310_000], minimum=200_000)
print(summary)
```
