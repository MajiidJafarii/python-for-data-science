<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">لیست</h2>
  <p>لیست ترتیبی و تغییرپذیر است و می‌تواند مقدارهای تکراری داشته باشد. برای مجموعه‌ای از آیتم‌ها که قرار است تغییر کنند، انتخاب مناسبی است.</p>
</div>

```python
# Create and update a list
tasks = ["email", "report", "meeting"]
tasks.append("backup")
tasks.insert(1, "review")
completed = tasks.pop(0)

print("Completed:", completed)
print("Remaining:", tasks)
print("First two:", tasks[:2])
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">تاپل</h2>
  <p>تاپل ترتیبی اما تغییرناپذیر است. برای داده‌های ثابت مانند مختصات مناسب است. بازکردن تاپل، مقدارها را هم‌زمان در چند متغیر قرار می‌دهد.</p>
</div>

```python
# Unpack a fixed coordinate
location = (35.6892, 51.3890)
latitude, longitude = location

print(f"Latitude: {latitude}")
print(f"Longitude: {longitude}")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">مجموعه یا set</h2>
  <p>مجموعه اعضای یکتا نگه می‌دارد و ترتیب آن مبنای قابل اتکایی ندارد. اجتماع، اشتراک و تفاضل مجموعه‌ها در مقایسهٔ گروه‌ها کاربردی است.</p>
</div>

```python
# Compare unique skill sets
team_a = {"python", "sql", "git"}
team_b = {"python", "excel", "git"}

print("Common:", team_a & team_b)
print("All skills:", team_a | team_b)
print("Only team A:", team_a - team_b)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">دیکشنری</h2>
  <p>دیکشنری داده را به‌صورت کلید و مقدار نگه می‌دارد. کلیدها یکتا هستند. متد <code>get</code> هنگام نبودن کلید، خطا ایجاد نمی‌کند.</p>
</div>

```python
# Store a product record
product = {
    "id": 101,
    "name": "Keyboard",
    "price": 1_850_000,
    "stock": 7,
}

product["stock"] -= 1
product["category"] = "Accessories"

print(product["name"])
print(product.get("discount", 0))
print(product)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ساختار تو در تو</h2>
  <p>در داده‌های واقعی معمولاً یک لیست از دیکشنری‌ها داریم؛ هر دیکشنری یک رکورد است.</p>
</div>

```python
# Analyze a list of records
orders = [
    {"id": 1, "customer": "Ali", "amount": 480_000},
    {"id": 2, "customer": "Sara", "amount": 720_000},
    {"id": 3, "customer": "Ali", "amount": 310_000},
]

total_amount = sum(order["amount"] for order in orders)
customers = {order["customer"] for order in orders}

print(f"Total: {total_amount:,}")
print("Customers:", customers)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Summarize a shopping cart
cart = [
    {"name": "Notebook", "price": 85_000, "quantity": 2},
    {"name": "Pen", "price": 25_000, "quantity": 4},
    {"name": "Notebook", "price": 85_000, "quantity": 1},
]

cart_total = sum(item["price"] * item["quantity"] for item in cart)
unique_products = {item["name"] for item in cart}

print(f"Cart total: {cart_total:,}")
print("Unique products:", unique_products)
```
