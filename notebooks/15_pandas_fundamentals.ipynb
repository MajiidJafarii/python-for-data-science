<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">Series و DataFrame</h2>
  <p>Series یک ستون برچسب‌دار و DataFrame یک جدول با سطر و ستون است. Pandas برای پاک‌سازی و تحلیل دادهٔ جدولی طراحی شده است.</p>
</div>

```python
# Create a DataFrame from records
import pandas as pd

orders = pd.DataFrame([
    {"order_id": 101, "city": "Tehran", "product": "Book", "quantity": 2, "unit_price": 180_000},
    {"order_id": 102, "city": "Shiraz", "product": "Pen", "quantity": 5, "unit_price": 35_000},
    {"order_id": 103, "city": "Tehran", "product": "Bag", "quantity": 1, "unit_price": 950_000},
    {"order_id": 104, "city": "Tabriz", "product": "Book", "quantity": 3, "unit_price": 180_000},
    {"order_id": 105, "city": "Shiraz", "product": "Bag", "quantity": 2, "unit_price": 950_000},
])

orders
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">بازرسی داده</h2>
  <p>پیش از تحلیل، شکل جدول، نوع ستون‌ها و چند ردیف اول را بررسی کنید. این مرحله بسیاری از خطاهای بعدی را آشکار می‌کند.</p>
</div>

```python
# Inspect structure and descriptive statistics
print("Shape:", orders.shape)
print("Columns:", orders.columns.tolist())
print("Data types:\n", orders.dtypes)
orders.select_dtypes(include="number").describe()
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">انتخاب با loc و iloc</h2>
  <p><code>loc</code> برچسب سطر و ستون را می‌گیرد؛ <code>iloc</code> بر اساس جایگاه عددی انتخاب می‌کند.</p>
</div>

```python
# Select columns and rows explicitly
print(orders.loc[:, ["order_id", "city", "product"]])
print(orders.iloc[:3, :2])
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">فیلتر و ساخت ستون</h2>
  <p>شرط یک Series بولی می‌سازد. برای چند شرط از <code>&</code> و <code>|</code> همراه پرانتز استفاده کنید.</p>
</div>

```python
# Add a calculated column and filter rows
orders = orders.assign(
    total=lambda frame: frame["quantity"] * frame["unit_price"]
)

large_orders = orders[
    (orders["total"] >= 500_000) & (orders["quantity"] >= 1)
]

large_orders.sort_values("total", ascending=False)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">گروه‌بندی</h2>
  <p><code>groupby</code> داده را به گروه‌ها تقسیم، محاسبه و نتیجه را ترکیب می‌کند. نام‌گذاری خروجی‌ها خوانایی گزارش را بیشتر می‌کند.</p>
</div>

```python
# Aggregate sales by city
city_summary = (
    orders.groupby("city", as_index=False)
    .agg(
        order_count=("order_id", "count"),
        total_sales=("total", "sum"),
        average_order=("total", "mean"),
    )
    .sort_values("total_sales", ascending=False)
)

city_summary
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">تمرین کوتاه</h2>
  <div style="background:#fff7df; border:1px solid #f0c36d; padding:12px; border-radius:8px;">برای هر محصول، تعداد سفارش، مجموع تعداد فروخته‌شده و درآمد کل را محاسبه کنید؛ سپس محصول پردرآمد را مشخص کنید.</div>
</div>

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Build a product-level sales report
product_summary = (
    orders.groupby("product", as_index=False)
    .agg(
        order_count=("order_id", "count"),
        units_sold=("quantity", "sum"),
        revenue=("total", "sum"),
    )
    .sort_values("revenue", ascending=False)
)

best_product = product_summary.iloc[0]["product"]
print("Best product:", best_product)
product_summary
```
