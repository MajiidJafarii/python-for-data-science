<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">تصادفی اما تکرارپذیر</h2>
  <p>برای آزمایش علمی باید نتیجه با seed ثابت تکرارپذیر باشد. روش جدید NumPy استفاده از <code>default_rng</code> است.</p>
</div>

```python
# Create a reproducible random generator
import numpy as np

rng = np.random.default_rng(seed=2026)
random_values = rng.random(5)
print(np.round(random_values, 4))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">شبیه‌سازی سکه و تاس</h2>
  <p>احتمال تجربی نسبت دفعات وقوع رویداد به تعداد آزمایش‌هاست. با افزایش تعداد آزمایش، مقدار تجربی معمولاً به احتمال نظری نزدیک‌تر می‌شود.</p>
</div>

```python
# Simulate fair coin flips and dice rolls
coin_flips = rng.choice(["H", "T"], size=10_000)
dice_rolls = rng.integers(1, 7, size=10_000)

probability_heads = np.mean(coin_flips == "H")
probability_six = np.mean(dice_rolls == 6)
probability_even = np.mean(dice_rolls % 2 == 0)

print("P(heads):", round(probability_heads, 4))
print("P(six):", round(probability_six, 4))
print("P(even):", round(probability_even, 4))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">آمار توصیفی</h2>
  <p>میانگین مرکز حسابی، میانه نقطهٔ وسط، واریانس پراکندگی مربعی و انحراف معیار پراکندگی در واحد اصلی داده را بیان می‌کند. صدک‌ها شکل توزیع را بهتر نشان می‌دهند.</p>
</div>

```python
# Summarize a numeric sample
response_times = np.array([120, 135, 128, 142, 119, 131, 450, 125, 138])

print("Mean:", round(response_times.mean(), 2))
print("Median:", np.median(response_times))
print("Variance:", round(response_times.var(ddof=1), 2))
print("Standard deviation:", round(response_times.std(ddof=1), 2))
print("Quartiles:", np.percentile(response_times, [25, 50, 75]))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">توزیع دوجمله‌ای</h2>
  <p>توزیع دوجمله‌ای تعداد موفقیت در n آزمایش مستقل با احتمال موفقیت ثابت p را مدل می‌کند؛ مثلاً تعداد خرید از میان بازدیدکنندگان.</p>
</div>

```python
# Simulate conversions from independent visitors
visitors_per_day = 100
conversion_probability = 0.08
simulated_days = 5_000

daily_conversions = rng.binomial(
    n=visitors_per_day,
    p=conversion_probability,
    size=simulated_days,
)

print("Expected conversions:", visitors_per_day * conversion_probability)
print("Simulated mean:", round(daily_conversions.mean(), 2))
print("P(at least 10):", round(np.mean(daily_conversions >= 10), 3))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">توزیع نرمال و استانداردسازی</h2>
  <p>بسیاری از اندازه‌گیری‌های طبیعی تقریباً زنگوله‌ای‌اند. استانداردسازی z-score فاصلهٔ هر مقدار از میانگین را بر حسب انحراف معیار بیان می‌کند.</p>
</div>

```python
# Generate normal data and calculate z-scores
exam_scores = rng.normal(loc=70, scale=10, size=1_000)
z_scores = (exam_scores - exam_scores.mean()) / exam_scores.std()

within_one_std = np.mean(np.abs(z_scores) <= 1)
print("Sample mean:", round(exam_scores.mean(), 2))
print("Sample standard deviation:", round(exam_scores.std(), 2))
print("Within one standard deviation:", round(within_one_std, 3))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">قانون اعداد بزرگ</h2>
  <p>میانگین تجمعی نشان می‌دهد با زیادشدن نمونه، برآورد میانگین به مقدار مورد انتظار نزدیک می‌شود.</p>
</div>

```python
# Observe convergence of the sample mean
rolls = rng.integers(1, 7, size=20_000)
cumulative_mean = np.cumsum(rolls) / np.arange(1, rolls.size + 1)

checkpoints = [10, 100, 1_000, 20_000]
for checkpoint in checkpoints:
    print(checkpoint, round(cumulative_mean[checkpoint - 1], 4))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">کوواریانس و همبستگی</h2>
  <p>کوواریانس جهت تغییر مشترک را نشان می‌دهد. همبستگی آن را به بازهٔ منفی یک تا یک مقیاس می‌کند. همبستگی به‌تنهایی رابطهٔ علت و معلولی را ثابت نمی‌کند.</p>
</div>

```python
# Measure association between advertising and sales
advertising = rng.uniform(10, 100, size=200)
sales = 3.2 * advertising + rng.normal(0, 30, size=200)

covariance_matrix = np.cov(advertising, sales, ddof=1)
correlation_matrix = np.corrcoef(advertising, sales)

print("Covariance:", round(covariance_matrix[0, 1], 2))
print("Correlation:", round(correlation_matrix[0, 1], 3))
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید و سپس با پاسخ پیشنهادی مقایسه کنید.</p>
</div>

```python
# Estimate pi with Monte Carlo simulation
sample_size = 200_000
points = rng.random((sample_size, 2))
squared_distance = points[:, 0] ** 2 + points[:, 1] ** 2
inside_circle = squared_distance <= 1
estimated_pi = 4 * inside_circle.mean()

print("Estimated pi:", round(estimated_pi, 5))
print("Absolute error:", round(abs(np.pi - estimated_pi), 5))
```
