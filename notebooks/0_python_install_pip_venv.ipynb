  <h2 style="color:#173f5f;">۱. نصب Python در macOS</h2>
  <p>در macOS بهتر است Python 3 را از <a href="https://www.python.org/downloads/">python.org</a> یا Homebrew نصب کنید. پس از نصب، Terminal را باز و نسخه و مسیر مفسر را بررسی کنید.</p>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Check Python and pip
python3 --version
python3 -m pip --version
which python3

# Homebrew alternative
brew install python</code></pre>
    <p>در ویندوز هنگام نصب گزینهٔ <code>Add Python to PATH</code> را فعال کنید و معمولاً از دستور <code>py</code> استفاده کنید.</p>
</div>

```python
# Inspect the interpreter used by this notebook
import platform
import sys

print("Python version:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Operating system:", platform.system())
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">۲. چرا python -m pip؟</h2>
  <p>دستور <code>pip</code> ممکن است به Python دیگری متصل باشد. نوشتن <code>python -m pip</code> تضمین می‌کند بسته برای همان مفسری نصب شود که انتخاب کرده‌اید.</p>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Upgrade pip
python3 -m pip install --upgrade pip

# Install packages
python3 -m pip install numpy pandas matplotlib scikit-learn

# Inspect installed packages
python3 -m pip list
python3 -m pip show pandas

# Upgrade or remove a package
python3 -m pip install --upgrade pandas
python3 -m pip uninstall package_name</code></pre>
</div>

```python
# Ask pip for its version without changing the environment
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "pip", "--version"],
    capture_output=True,
    text=True,
    check=True,
)
print(result.stdout.strip())
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">۳. محیط مجازی چیست؟</h2>
  <p>محیط مجازی یک پوشهٔ مستقل برای مفسر و بسته‌های هر پروژه است. با این کار نسخهٔ کتابخانه‌های پروژه‌ها با هم تداخل پیدا نمی‌کنند. نام رایج پوشه <code>.venv</code> است و نباید وارد Git شود.</p>
    <h3>macOS و Linux</h3>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Create a virtual environment
python3 -m venv .venv

# Activate it
source .venv/bin/activate

# Verify the active interpreter
which python
python -m pip --version

# Leave the environment
deactivate</code></pre>
    <h3>Windows PowerShell</h3>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Create and activate a virtual environment
py -m venv .venv
.\.venv\Scripts\Activate.ps1

# Leave the environment
deactivate</code></pre>
</div>

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">۴. نصب وابستگی‌های پروژه</h2>
  <p>پس از فعال‌سازی محیط، بسته‌ها را نصب و فهرست نسخه‌ها را ثبت کنید. برای پروژهٔ آموزشی بهتر است ابتدا بسته‌ها را بدون محدودیت نصب و سپس خروجی دقیق محیط سالم را ذخیره کنید.</p>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Install project dependencies
python -m pip install numpy pandas matplotlib scikit-learn jupyterlab ipykernel

# Save exact installed versions
python -m pip freeze > requirements.txt

# Rebuild the same environment later
python -m pip install -r requirements.txt</code></pre>
</div>

```python
# Check whether the main course packages are available
from importlib.util import find_spec

package_names = ["numpy", "pandas", "matplotlib", "sklearn"]
availability = {
    package: find_spec(package) is not None
    for package in package_names
}
print(availability)
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">۵. اتصال venv به Jupyter</h2>
  <p>اگر Jupyter با محیط دیگری اجرا شود، ممکن است بستهٔ نصب‌شده را پیدا نکند. محیط پروژه را به‌عنوان kernel ثبت کنید و سپس آن را از منوی Kernel انتخاب کنید.</p>
    <pre dir="ltr" style="text-align:left; background:#f6f8fa; padding:12px; border-radius:8px;"><code># Run these commands inside the active environment
python -m pip install ipykernel
python -m ipykernel install --user --name python-review --display-name "Python Review"

# Start JupyterLab
python -m pip install jupyterlab
python -m jupyter lab

# List and remove kernels when needed
jupyter kernelspec list
jupyter kernelspec uninstall python-review</code></pre>
</div>

  <h2 style="color:#173f5f;">۶. خطاهای رایج</h2>
  <ul>
  <li><b>ModuleNotFoundError:</b> بسته را در همان مفسر kernel نصب کنید.</li>
  <li><b>Permission denied:</b> محیط مجازی بسازید؛ از <code>sudo pip</code> استفاده نکنید.</li>
  <li><b>pip قدیمی:</b> دستور <code>python -m pip install --upgrade pip</code> را اجرا کنید.</li>
  <li><b>فعال نشدن venv:</b> مسیر فعال‌سازی را با سیستم‌عامل تطبیق دهید.</li>
  <li><b>انتخاب Python اشتباه:</b> مقدار <code>sys.executable</code> را داخل Notebook بررسی کنید.</li>
</ul>
