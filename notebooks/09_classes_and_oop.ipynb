<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">کلاس و شیء</h2>
  <p>کلاس نقشهٔ ساخت شیء است. هر شیء وضعیت خود را در ویژگی‌ها و رفتار خود را در متدها نگه می‌دارد. <code>self</code> به همان نمونه اشاره می‌کند.</p>
</div>

```python
# Model a bank account with state and behavior
class BankAccount:
    bank_name = "Example Bank"

    def __init__(self, owner, initial_balance=0):
        if initial_balance < 0:
            raise ValueError("initial balance cannot be negative")
        self.owner = owner
        self._balance = initial_balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self._balance += amount
        return self._balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("insufficient balance")
        self._balance -= amount
        return self._balance

    def __repr__(self):
        return f"BankAccount(owner={self.owner!r}, balance={self.balance})"

account = BankAccount("Sara", 1_000_000)
account.deposit(250_000)
account.withdraw(180_000)
print(account)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ویژگی کلاسی و نمونه‌ای</h2>
  <p><code>bank_name</code> میان همهٔ نمونه‌ها مشترک است؛ اما <code>owner</code> و <code>_balance</code> برای هر نمونه جدا هستند. علامت زیرخط یک قرارداد برای استفادهٔ داخلی است.</p>
</div>

```python
# Compare independent instances
first_account = BankAccount("Ali", 500_000)
second_account = BankAccount("Nima", 900_000)

first_account.deposit(100_000)
print(first_account.balance)
print(second_account.balance)
print(BankAccount.bank_name)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">وراثت</h2>
  <p>وراثت زمانی مفید است که رابطهٔ «نوعی از» وجود دارد. کلاس فرزند می‌تواند رفتار والد را استفاده یا بازنویسی کند.</p>
</div>

```python
# Extend an existing account type
class SavingsAccount(BankAccount):
    def __init__(self, owner, initial_balance=0, interest_rate=0.05):
        super().__init__(owner, initial_balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self._balance += interest
        return interest

savings = SavingsAccount("Mina", 2_000_000, interest_rate=0.08)
earned_interest = savings.add_interest()
print(f"Interest: {earned_interest:,.0f}")
print(savings)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">dataclass برای کلاس‌های داده‌محور</h2>
  <p>اگر کلاس بیشتر برای نگه‌داری داده است، <code>dataclass</code> متدهای تکراری مانند سازنده و نمایش را خودکار می‌سازد.</p>
</div>

```python
# Define a concise data-focused class
from dataclasses import dataclass

@dataclass
class Product:
    name: str
    price: float
    stock: int = 0

    def inventory_value(self) -> float:
        return self.price * self.stock

product = Product("Monitor", 8_500_000, 4)
print(product)
print(f"Inventory value: {product.inventory_value():,.0f}")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Manage course enrollment rules
class Course:
    def __init__(self, title, capacity):
        if capacity <= 0:
            raise ValueError("capacity must be positive")
        self.title = title
        self.capacity = capacity
        self.students = []

    def enroll(self, student):
        if student in self.students:
            return False
        if len(self.students) >= self.capacity:
            return False
        self.students.append(student)
        return True

    @property
    def remaining_seats(self):
        return self.capacity - len(self.students)

course = Course("Python", 2)
print(course.enroll("Sara"))
print(course.enroll("Ali"))
print(course.enroll("Nima"))
print("Remaining seats:", course.remaining_seats)
```
