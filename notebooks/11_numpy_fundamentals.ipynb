<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">چرا NumPy؟</h2>
  <p>NumPy آرایهٔ چندبعدی سریع و ابزارهای محاسبات عددی ارائه می‌کند. برخلاف لیست پایتون، اعضای آرایه معمولاً یک نوع داده دارند.</p>
</div>

```python
# Import NumPy with its standard alias
import numpy as np

values = np.array([12, 15, 18, 21], dtype=np.float64)
print(values)
print("Shape:", values.shape)
print("Dimensions:", values.ndim)
print("Data type:", values.dtype)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ساخت آرایه</h2>
  <p><code>zeros</code>، <code>ones</code>، <code>arange</code> و <code>linspace</code> آرایه‌های رایج را بدون نوشتن تک‌تک اعضا می‌سازند.</p>
</div>

```python
# Create arrays with common constructors
print(np.zeros((2, 3)))
print(np.ones(4, dtype=int))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">اندیس و برش</h2>
  <p>در آرایهٔ دوبعدی، بعد اول ردیف و بعد دوم ستون است. برش‌ها معمولاً view می‌سازند؛ بنابراین تغییر آن‌ها می‌تواند آرایهٔ اصلی را تغییر دهد.</p>
</div>

```python
# Select rows, columns, and sub-arrays
matrix = np.array([
    [10, 12, 14],
    [16, 18, 20],
    [22, 24, 26],
])

print("Single item:", matrix[1, 2])
print("First row:", matrix[0])
print("Second column:", matrix[:, 1])
print("Top-left block:\n", matrix[:2, :2])
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">محاسبهٔ برداری و Broadcasting</h2>
  <p>عملگر روی کل آرایه اجرا می‌شود. Broadcasting اجازه می‌دهد آرایه‌های سازگار با شکل متفاوت، بدون تکرار دستی با هم محاسبه شوند.</p>
</div>

```python
# Apply vectorized operations
prices = np.array([120_000, 85_000, 240_000, 60_000])
discounted = prices * 0.9

monthly_sales = np.array([
    [120, 140, 135],
    [90, 110, 105],
])
growth_factors = np.array([1.0, 1.05, 1.10])
projected = monthly_sales * growth_factors

print(discounted)
print(projected)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">فیلتر و آمار</h2>
  <p>شرط روی آرایه یک ماسک Boolean می‌سازد. توابع تجمیعی مانند میانگین، مجموع، کمینه و بیشینه کل داده را خلاصه می‌کنند.</p>
</div>

```python
# Filter and summarize numeric data
scores = np.array([12, 17.5, 8, 19, 15, 6.5])
passed = scores[scores >= 10]

print("Passed:", passed)
print("Count:", passed.size)
print("Average:", passed.mean())
print("Highest:", scores.max())
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Analyze weekly temperatures
temperatures_c = np.array([18.5, 21, 23.5, 19, 25, 24, 20])
average_c = temperatures_c.mean()
above_average = temperatures_c[temperatures_c > average_c]
temperature_range = np.ptp(temperatures_c)
temperatures_f = temperatures_c * 9 / 5 + 32

print("Average C:", round(average_c, 2))
print("Above average:", above_average)
print("Range:", temperature_range)
print("Fahrenheit:", temperatures_f)
```
