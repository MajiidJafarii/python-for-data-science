<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">دیتاست Iris</h2>
  <p>Iris شامل ۱۵۰ نمونه از سه گونهٔ گل زنبق و چهار اندازه‌گیری طول و عرض کاسبرگ و گلبرگ است. این داده داخل scikit-learn قرار دارد و برای بارگذاری آن به دانلود اینترنتی نیاز نیست.</p>
</div>

```python
# Load the built-in Iris dataset as a pandas object
import pandas as pd
from sklearn.datasets import load_iris

iris_bundle = load_iris(as_frame=True)
iris = iris_bundle.frame.copy()

print("Shape:", iris.shape)
print("Target names:", iris_bundle.target_names)
iris.head()
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">نام‌گذاری خوانا</h2>
  <p>نام ستون‌های اولیه واحد و فاصله دارند. برای کدنویسی تمیز، نام‌ها را کوتاه می‌کنیم و شمارهٔ target را به نام گونه تبدیل می‌کنیم.</p>
</div>

```python
# Rename columns and map numeric targets to species names
iris = iris.rename(columns={
    "sepal length (cm)": "sepal_length",
    "sepal width (cm)": "sepal_width",
    "petal length (cm)": "petal_length",
    "petal width (cm)": "petal_width",
})
iris["species"] = iris["target"].map(
    dict(enumerate(iris_bundle.target_names))
)
iris = iris.drop(columns="target")

print(iris.dtypes)
iris.head()
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">بررسی کیفیت و توصیف</h2>
  <p>پیش از تحلیل، اندازه، نوع ستون‌ها، مقدار گمشده، تکرار و آمار توصیفی را بررسی می‌کنیم.</p>
</div>

```python
# Profile the dataset before analysis
print("Rows and columns:", iris.shape)
print("Missing values:\n", iris.isna().sum())
print("Duplicate rows:", iris.duplicated().sum())
iris.describe().round(2)
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">فیلتر با query</h2>
  <p><code>query</code> برای شرط‌های خوانا مناسب است. در نام ستون‌های استانداردشده دیگر نیازی به کوتیشن خاص نداریم.</p>
</div>

```python
# Select flowers with relatively large petals
large_petals = (
    iris.query("petal_length > 5 and petal_width >= 1.8")
    .sort_values(["petal_length", "petal_width"], ascending=False)
)

print("Selected rows:", len(large_petals))
large_petals.head()
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">گروه‌بندی بر اساس گونه</h2>
  <p>میانگین و انحراف معیار هر ویژگی برای هر گونه، تفاوت گروه‌ها را روشن می‌کند.</p>
</div>

```python
# Compare feature summaries across species
species_summary = (
    iris.groupby("species")
    .agg(
        sample_count=("species", "size"),
        mean_sepal_length=("sepal_length", "mean"),
        mean_petal_length=("petal_length", "mean"),
        std_petal_length=("petal_length", "std"),
        mean_petal_width=("petal_width", "mean"),
    )
    .round(2)
)

species_summary
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">همبستگی ویژگی‌ها</h2>
  <p>ماتریس همبستگی رابطهٔ خطی جفت ویژگی‌ها را نشان می‌دهد. در Iris، طول و عرض گلبرگ معمولاً همبستگی بالایی دارند.</p>
</div>

```python
# Calculate feature correlations
numeric_columns = iris.select_dtypes(include="number")
correlations = numeric_columns.corr().round(3)
correlations
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">تمرین پاک‌سازی واقعی</h2>
  <p>برای تمرین، چند مقدار را عمداً حذف می‌کنیم و سپس هر مقدار گمشده را با میانهٔ همان گونه پر می‌کنیم. این روش تفاوت طبیعی گروه‌ها را بهتر از میانهٔ کل حفظ می‌کند.</p>
</div>

```python
# Introduce and fill grouped missing values
iris_missing = iris.copy()
iris_missing.loc[[5, 55, 105], "sepal_width"] = pd.NA

group_medians = iris_missing.groupby("species")["sepal_width"].transform("median")
iris_missing["sepal_width"] = iris_missing["sepal_width"].fillna(group_medians)

print("Remaining missing values:", iris_missing["sepal_width"].isna().sum())
print(iris_missing.loc[[5, 55, 105], ["species", "sepal_width"]])
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">ذخیره و بازیابی CSV</h2>
  <p>در پروژهٔ واقعی ممکن است داده از CSV یا Excel وارد شود. در این مثال فایل موقت می‌سازیم تا نوت‌بوک پس از اجرا فایل اضافی باقی نگذارد.</p>
</div>

```python
# Round-trip the cleaned dataset through a temporary CSV file
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as temp_dir:
    csv_path = Path(temp_dir) / "iris_clean.csv"
    iris_missing.to_csv(csv_path, index=False, encoding="utf-8")
    reloaded_iris = pd.read_csv(csv_path)

print("Reloaded shape:", reloaded_iris.shape)
print("Same columns:", reloaded_iris.columns.tolist() == iris_missing.columns.tolist())
```

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">تمرین</h2>
  <div style="background:#fff7df; border:1px solid #f0c36d; padding:12px; border-radius:8px;">ستون <code>petal_ratio = petal_length / petal_width</code> را بسازید؛ سپس میانگین این نسبت را برای هر گونه محاسبه و گونه‌ها را نزولی مرتب کنید.</div>
</div>

<div dir="rtl" style="text-align:right; font-family:Tahoma,Arial,sans-serif; line-height:1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید و سپس با پاسخ پیشنهادی مقایسه کنید.</p>
</div>

```python
# Engineer a ratio feature and summarize it by species
iris_with_ratio = iris.assign(
    petal_ratio=lambda frame: frame["petal_length"] / frame["petal_width"]
)
ratio_summary = (
    iris_with_ratio.groupby("species", as_index=False)
    .agg(mean_petal_ratio=("petal_ratio", "mean"))
    .sort_values("mean_petal_ratio", ascending=False)
)
ratio_summary["mean_petal_ratio"] = ratio_summary["mean_petal_ratio"].round(2)

ratio_summary
```
