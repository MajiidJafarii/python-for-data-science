<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">حلقهٔ for</h2>
  <p>حلقهٔ <code>for</code> روی اعضای یک iterable حرکت می‌کند. برای شمارندهٔ خوانا از <code>enumerate</code> استفاده کنید.</p>
</div>

```python
# Enumerate tasks from one
tasks = ["Collect data", "Clean data", "Create report"]

for number, task in enumerate(tasks, start=1):
    print(f"{number}. {task}")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">range و zip</h2>
  <p><code>range</code> دنبالهٔ عددی می‌سازد و <code>zip</code> چند مجموعه را عضو به عضو کنار هم قرار می‌دهد.</p>
</div>

```python
# Pair names with scores
names = ["Ali", "Sara", "Nima"]
scores = [16, 19, 14]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

print("Even numbers:", list(range(2, 11, 2)))
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">break و continue</h2>
  <p><code>continue</code> ادامهٔ همان دور را رد می‌کند و <code>break</code> کل حلقه را متوقف می‌کند.</p>
</div>

```python
# Find the first valid positive value
values = [None, -2, 0, 7, 10]

for value in values:
    if value is None:
        continue
    if value > 0:
        print("First positive value:", value)
        break
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">حلقهٔ while</h2>
  <p>وقتی تعداد تکرار از قبل مشخص نیست، <code>while</code> مناسب است. همیشه مطمئن شوید شرط حلقه سرانجام نادرست می‌شود.</p>
</div>

```python
# Reduce a balance with a safe loop condition
balance = 350_000
payment = 100_000
step = 0

while balance > 0:
    step += 1
    paid = min(payment, balance)
    balance -= paid
    print(step, paid, balance)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">Comprehension</h2>
  <p>برای تبدیل یا فیلتر ساده، comprehension از حلقهٔ چندخطی جمع‌وجورتر است. منطق پیچیده را در تابع جداگانه نگه دارید.</p>
</div>

```python
# Transform and filter values
prices = [120_000, 85_000, 240_000, 60_000]
discounted = [price * 0.9 for price in prices]
expensive = [price for price in prices if price >= 100_000]

print(discounted)
print(expensive)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Summarize valid daily sales
daily_sales = [450_000, 0, 720_000, -1, 610_000, 530_000]
valid_sales = [sale for sale in daily_sales if sale > 0]

sales_total = sum(valid_sales)
sales_average = sales_total / len(valid_sales)
best_sale = max(valid_sales)

print(f"Total: {sales_total:,}")
print(f"Average: {sales_average:,.0f}")
print(f"Best day: {best_sale:,}")
```
