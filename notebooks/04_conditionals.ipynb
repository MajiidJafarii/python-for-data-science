<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ساختار شرط</h2>
  <p>بلوک شرط با تورفتگی مشخص می‌شود. فقط نخستین شاخه‌ای که شرط آن درست باشد اجرا خواهد شد.</p>
</div>

```python
# Classify a score
score = 82

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "Needs improvement"

print(grade)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">ترکیب شرط‌ها</h2>
  <p>با <code>and</code> باید همهٔ شرط‌ها درست باشند؛ با <code>or</code> درست‌بودن یکی کافی است. پرانتز، خوانایی و اولویت را روشن می‌کند.</p>
</div>

```python
# Check access rules
role = "editor"
is_active = True
attempts = 1

can_login = is_active and attempts < 3
can_publish = can_login and role in {"admin", "editor"}

print("Can login:", can_login)
print("Can publish:", can_publish)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">Truthy و Falsy</h2>
  <p>مقدارهایی مانند صفر، رشتهٔ خالی، لیست خالی و <code>None</code> در شرط نادرست تلقی می‌شوند. این ویژگی شرط‌ها را کوتاه و خوانا می‌کند.</p>
</div>

```python
# Use truthy and falsy values
search_query = ""
results = []

if not search_query:
    print("Enter a search phrase.")

if results:
    print("Results are available.")
else:
    print("No results found.")
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">عبارت شرطی کوتاه</h2>
  <p>برای انتخاب ساده بین دو مقدار می‌توان از عبارت شرطی یک‌خطی استفاده کرد؛ اما شرط‌های پیچیده را کامل بنویسید.</p>
</div>

```python
# Select a label with a conditional expression
stock = 4
stock_label = "Available" if stock > 0 else "Out of stock"
print(stock_label)
```

<div dir="rtl" style="text-align: right; font-family: Tahoma, Arial, sans-serif; line-height: 1.9;">
  <h2 style="color:#173f5f;">پاسخ پیشنهادی</h2>
  <p>ابتدا پاسخ خودتان را بنویسید؛ سپس این سلول را اجرا و مقایسه کنید.</p>
</div>

```python
# Calculate shipping with business rules
order_total = 720_000
is_premium = True
standard_shipping = 80_000

if order_total >= 1_000_000:
    shipping_cost = 0
elif is_premium:
    shipping_cost = standard_shipping * 0.5
else:
    shipping_cost = standard_shipping

final_amount = order_total + shipping_cost
print(f"Shipping: {shipping_cost:,.0f}")
print(f"Final amount: {final_amount:,.0f}")
```
